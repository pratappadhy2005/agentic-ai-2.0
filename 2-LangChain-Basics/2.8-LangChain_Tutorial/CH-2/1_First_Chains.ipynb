# Chains In LangChain


```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()  # This loads the variables from .env

llm_openai = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

# ***First Chain***

```python
# Task -1 [Prompt]
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("human", "{input}"),
])
```

```python
# Task -2 [LLM]

llm_openai = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

```python
# Task -3 [String Parser]

str_parser = StrOutputParser()
```

### ***Manual Invocation***

```python
template = prompt_template.invoke({"input": "What is the capital of France?"})
llm_response = llm_openai.invoke(template)
parsed_response = str_parser.invoke(llm_response)
print(parsed_response)
```

# ***Chaining Invocation Using Pipe***

```python
chain = prompt_template | llm_openai | str_parser
response = chain.invoke({"input": "What is the capital of France?"})
print(response)
```

# ***Chaining with RunnableSequence***

```python
from langchain_core.runnables import RunnableSequence
chain_1 = RunnableSequence(prompt_template, llm_openai, str_parser)
response = chain_1.invoke({"input": "What is the capital of France?"})
print(response)
```
