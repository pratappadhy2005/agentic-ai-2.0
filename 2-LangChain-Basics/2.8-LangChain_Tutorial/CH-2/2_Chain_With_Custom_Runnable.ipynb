# ***Chaining Using Custom Runnable***

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()  # This loads the variables from .env

llm_openai = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

```python
# Task -1 [Prompt]
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("human", "{input}"),
])
```

```python
# Task -2 [LLM]

llm_openai = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

```python
# Task -3 [String Parser]

str_parser = StrOutputParser()
```

```python
# Task -4 [Custom Runnable]
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text: str) -> dict:
    return {"text": text}

# Create a RunnableLambda for the dictionary_maker function
dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
# Task -5 [Template for post]
prompt_post = ChatPromptTemplate.from_messages(
    messages = [
        ("system", "You are a social media generator."),
        ("human", "Create a post for the following text for the social media platform {text}")
    ]
)
```

```python
# Task -5 [Generate the post for the given text]

llm_openai_post = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

```python
# Task -6 [Output the post in the given format]

str_parser_post = StrOutputParser()
```

```python
# Chain

chain = prompt_template | llm_openai | str_parser | dictionary_maker_runnable | prompt_post | llm_openai_post | str_parser_post
chain.invoke("What is the capital of France?")
```
