# ***Messages***


```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
import os

from dotenv import load_dotenv

load_dotenv()  # This loads the variables from .env

llm_openai = ChatOpenAI(model_name="gpt-5-mini", temperature=0, openai_api_key=os.environ.get("OPENAI_API_KEY"))

my_message = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="What is the capital of France?")
]

response = llm_openai.invoke(my_message)
print(response.content)

```

# **Prompts**

```python
my_message = [
    SystemMessage(content="You are a gen Z assistant who always answers in a fun way."),
    HumanMessage(content="Bro tell me about a fun fact about gen Z."),
]
response = llm_openai.invoke(my_message).content
print(response)
```

# ***PROMPT TEMPLATE***

```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter a topic: ")

dynamic_prompt = PromptTemplate.from_template(
    "Write a fun fact about {topic}."
)

ready_prompt = dynamic_prompt.invoke({"topic": user_input})

llm_response = llm_openai.invoke(ready_prompt).content
print(llm_response)
```

# ***Chat Prompt Template***

```python
from langchain_core.prompts import ChatPromptTemplate

user_input = input("Enter a topic: ")

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a gen Z assistant who always answers in a fun way."),
    ("user", "{input}"),
])

llm_response = llm_openai.invoke(prompt_template.invoke({"input": user_input})).content
print(llm_response)
```
