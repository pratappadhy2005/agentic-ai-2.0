### Chat Prompt Template

```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv

load_dotenv()

if os.getenv("OPENAI_API_KEY") is None:
    raise ValueError("OPENAI_API_KEY environment variable is not set")
else:
    print("OPENAI_API_KEY environment variable is set")
```

```python
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(
    model_name="gpt-5-nano",
    temperature=0.7
)

response = llm.invoke([HumanMessage(content="Hello, how are you?")])

print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

my_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that helps to create and curate posts on a social media platform."),
    ("human", "I want to create a post about the {topic}.")
])
```

```python
prompt_filled =my_prompt.invoke({"topic": "latest AI advancements"})
```

```python
response = llm.invoke(prompt_filled)
print(response.content)
```

### ***CHAINING WITH PROMPTS***

```python
chain = my_prompt | llm
response = chain.invoke({"topic": "latest AI advancements"})
print(response.content)
```
